```python
import pandas as pd
```

```python
# Load the World Happiness Report data
df_happy = pd.read_excel('DataForTable2.1WHR2023.xls')

# Load the Economic Freedom Index data
df_free = pd.read_excel('economic-freedom-of-the-world-2024-master-index-data-for-researchers-iso.xlsx', header=4)
```

```python
# Drop 'ISO Code 2' column and rename 'ISO Code 3' to 'Country Name'
df_free = df_free.drop(columns=['ISO Code 2'])
df_free = df_free.rename(columns={'ISO Code 3': 'Country Name'})
```

```python
df_happy.head()
```

```python
df_free.head()
```

```python
req_columns_free = ['Year','Countries',' Economic Freedom Summary Index']
df_free = df_free[req_columns_free]
```

```python
df_free.head()
```

```python
req_columns_happy = ['Country name','year','Life Ladder','Healthy life expectancy at birth']
df_happy = df_happy[req_columns_happy]
```

```python
df_happy.head()
```

```python
# Display rows with missing 'Year' values
rows_with_nan = df_free[df_free['Year'].isna()]
print("Rows with missing 'Year':")
print(rows_with_nan)

# Drop rows with missing 'Year'
df_free = df_free.dropna(subset=['Year'])
```

```python
df_free['Year'] = df_free['Year'].astype(int)
```

```python
df_free.head()
```

```python
df_happy.head()
```

```python
merged_df = pd.merge(df_free,df_happy,left_on=['Year','Countries'],right_on=['year','Country name'],how='inner')
```

```python
# Display first few rows of the merged dataset
merged_df.head()
```

```python
merged_df.columns
```

```python
merged_df.drop(columns=['Country name','year'],inplace=True)
merged_df.head()
```

```python
merged_df.rename(columns={' Economic Freedom Summary Index':'Economic Freedom Index'},inplace=True)
```

```python
eco_hap_df = merged_df.iloc[:]
eco_hap_df
```

```python
eco_hap_df.isna().sum()
```

```python
eco_hap_df[eco_hap_df['Economic Freedom Index'].isna()]
```

```python
eco_hap_df.dropna(inplace=True)
```

```python
eco_hap_df.isna().sum()
```

```python
eco_hap_df_2022 = eco_hap_df[eco_hap_df['Year']==2022]
```

```python
eco_hap_df_2022
```

```python
eco_hap_df_2022_sorted = eco_hap_df_2022.sort_values(by='Economic Freedom Index',ascending=False)
eco_hap_df_2022_sorted
```

```python
eco_hap_df_top_20 = eco_hap_df_2022_sorted.iloc[:20,:]
```

```python
eco_hap_df_top_20.reset_index(drop=True,inplace=True)
eco_hap_df_top_20
```
