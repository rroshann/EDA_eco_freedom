```python
# Importing pandas
import pandas as pd
```

```python
# Load the Economic Freedom Index data
freedom_index_df = pd.read_excel('economic-freedom-of-the-world-2024-master-index-data-for-researchers-iso.xlsx', header=4)
```

```python
freedom_index_df.head()
```

```python
# Drop 'ISO Code 2' column and rename 'ISO Code 3' to 'Country Name'
freedom_index_df = freedom_index_df.drop(columns=['ISO Code 2'])
freedom_index_df = freedom_index_df.rename(columns={'ISO Code 3': 'Country Name'})
```

```python
# Making sure we only have the required columns for the base dataset i.e Economic Freedom Index
req_columns_free = ['Year','Countries',' Economic Freedom Summary Index']
freedom_index_df = freedom_index_df[req_columns_free]
```

```python
freedom_index_df.head()
```

```python
# Rename the ' Economic Freedom Summary Index' to 'Economic Freedom Index'
freedom_index_df = freedom_index_df.rename(columns={' Economic Freedom Summary Index':'Economic Freedom Index'})
```

```python
# Display rows with missing 'Year' values
rows_with_nan = freedom_index_df[freedom_index_df['Year'].isna()]
print("Rows with missing 'Year':")
print(rows_with_nan)

# Drop rows with missing 'Year'
freedom_index_df = freedom_index_df.dropna(subset=['Year'])
```

```python
# Change the datatype of Year column to int
freedom_index_df['Year'] = freedom_index_df['Year'].astype(int)
```

```python
# Filter the DataFrame for the specified year range i.e 2015-2022

freedom_index_df = freedom_index_df[(freedom_index_df['Year']>=2015) & (freedom_index_df['Year']<=2022)]
```

```python
freedom_index_df.head()
```

```python
# Checking the number of null values in the base dataset
freedom_index_df.isna().sum()
```

```python
# Checking what and where the missing values are
freedom_index_df[freedom_index_df['Economic Freedom Index'].isna()]
```

```python
# Dropping the rows where there are null values
freedom_index_df.dropna(inplace=True)
```

```python
freedom_index_df.head()
```

```python
# Sort the dataset so that it looks neat and organized
freedom_index_df = freedom_index_df.sort_values(by=['Year','Countries']).reset_index(drop=True)
```

```python
freedom_index_df.head()
```

```python
# Group by 'Year' and find the country with the highest 'Economic Freedom Index' for each year
highest_freedom_index_df = freedom_index_df.loc[freedom_index_df.groupby('Year')['Economic Freedom Index'].idxmax()]

# Reset the index for better readability
highest_freedom_index_df = highest_freedom_index_df.reset_index(drop=True)

highest_freedom_index_df
```

```python
# Renaming Hong Kong for better readibility in the highest_freedom_index_df
highest_freedom_index_df['Countries'] = highest_freedom_index_df['Countries'].replace('Hong Kong SAR, China', 'Hong Kong')
```

```python
highest_freedom_index_df
```

```python
# Renaming Hong Kong for better readibility in the freedom_index_df
freedom_index_df['Countries'] = freedom_index_df['Countries'].replace('Hong Kong SAR, China', 'Hong Kong')
```

```python
# Sort the top 15 countries with the highest 'Economic Freedom Index' for the year 2022
freedom_index_df_top_15_2022 = freedom_index_df[freedom_index_df['Year']==2022].sort_values(by='Economic Freedom Index',ascending=False).iloc[:15].reset_index(drop=True)
```

```python
freedom_index_df_top_15_2022
```

```python
# Storing the top 15 countries in a list
top_15_countries = list(freedom_index_df_top_15_2022['Countries'])
top_15_countries
```

```python
# Write the freedom_index_df dataframe to a csv file named 'freedom_index.csv'
freedom_index_df.to_csv('freedom_index.csv', index=False)
```
